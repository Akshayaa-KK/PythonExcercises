### Use Case 2: Salary Processing System (HR / Payroll Automation)
Demonstrates input/output functions, default arguments, arbitrary args, returning values, and composing multiple functions.
What Problem It Solves:
Companies calculate employee salary differently depending on bonus percentage, incentives, PF, tax, etc.
How Function-Based Programming Helps:
We create multiple small reusable functions (bonus calc, tax calc, net salary calc)
Then compose them together.


```python
def get_bonusamount(base_salary,bonus_percentage):
    return base_salary * (bonus_percentage /100)

def get_tax_amount(base_salary):
    if base_salary >= 1200000:
        return base_salary * (12/100)
    return  0

def get_pf_amount(base_salary):
    return  base_salary * (5/100)


def salary_calculator(base_salary,bonuspercentage):
    bonus_amount = get_bonusamount(base_salary,bonuspercentage)
    tax_amount = get_tax_amount(base_salary)
    pf_amount = get_pf_amount(base_salary)
    net_salary = (base_salary + bonus_amount) - tax_amount - pf_amount
    return net_salary

net_salary = salary_calculator(500000,10)
print(f"The net salary is {net_salary}")
```

### Usecase3: Number Utility Tool
Create functions:
is_even(num) → returns True/False
find_max(*numbers) → returns highest number from arguments
perform_operation(num, operation) →
If operation = "square" → return num*num
If operation = "cube" → return numnumnum
If unknown → return "Invalid operation"


Test the functions with at least 5 numbers.


```python
def odd_or_even(num):
    if num % 2 == 0:
        return True
    return False

def find_max_num(num_list):
    if not num_list:
        return 0
    highest_num = num_list[0]

    for i in range(1,len(num_list)):
        if num_list[i] > highest_num:
            highest_num = num_list[i]
    return highest_num

def perform_operation(num,operation):
    if operation == "square":
        return num * 2
    elif operation == "cube":
        return num ** 3
    else:
        return None




print(odd_or_even(2))
print(find_max_num([1,2,3,4,5]))
print(perform_operation(2,"square"))
print(perform_operation(3,"cube"))
```
